## 1. Main Result

### Theorem 1.1, in the paper's stated form
For every viscosity nu > 0, the paper constructs a smooth external force f, compactly supported in space and time, and smooth incompressible velocity/pressure fields (u,p) on R^3 x [0,1) such that: (i) u and p have support in a common compact spatial set at each t<1; (ii) u(.,0)=0; (iii) kinetic energy remains uniformly bounded for 0<=t<1; and (iv) limsup as t approaches 1 of ||u(t)||_L-infinity is infinite. The paper further claims there is no global smooth solution with the same force and initial datum whose kinetic energy is uniformly bounded. It states that the result establishes alternatives (C) and (D) in Fefferman's formulation; compact support yields the periodic construction on T^3 as well.

### Important scope distinctions
- This is a **forced** Navier-Stokes construction. It is not a claim that unforced smooth initial data blow up.
- The initial velocity is zero; the force is nonzero and is part of the theorem's data.
- The theorem is for every positive viscosity. The zero-viscosity Euler case is not covered by Theorem 1.1.
- The constructed solution is smooth before t=1, has bounded L^2 kinetic energy, and loses bounded L-infinity velocity at the singular time. Do not conflate these different norms.

## 2. Governing Equation and Proof Objective

The forced incompressible equations used in Theorem 1.1 are

$$\partial_t u + (u\cdot\nabla)u - \nu\Delta u + \nabla p = f, \qquad \nabla\cdot u=0, \qquad u(\cdot,0)=0. \tag{1.1}$$

For a proposed velocity and pressure, define the momentum residual (the outline first sets nu=1) by

$$R(u,p)=\partial_t u+(u\cdot\nabla)u-\Delta u+\nabla p. \tag{3.1}$$

The central construction problem is not merely to make velocity large. It is to construct u,p with the desired velocity growth while arranging for the residual to extend as a smooth force through the singular time. The terms composing R may individually grow; the proof engineers cancellation in their sum and in every fixed Cartesian space-time derivative. The localized force is the completed residual.

### Proof target
Construct a solution smooth on [0,1), incompressible and compactly supported in space, with zero initial data, bounded kinetic energy, unbounded L-infinity velocity as t increases to 1, and a smooth compactly supported force. Then show that a global smooth bounded-energy solution with the same force and datum would have to agree before t=1 and hence inherit the unbounded growth, a contradiction.

## 3. Geometric Ansatz and Similarity Scales

The leading flow is axisymmetric in ordinary cylindrical coordinates (r, theta, z), centered on the vertical axis. It combines inward radial motion, azimuthal swirl, and axial outflow. A slightly asymmetric axial profile is used to retain the needed shear near the central dividing layer. The radial core width contracts faster than its axial length.

Let tau=1-t and choose a fixed h with 0<h<1/100. Define A=1/2+h and D=1/2-h. The similarity variables in equation (3.2) are

$$\tau=q(1-\eta^2), \qquad z=q^D\eta, \qquad X=\frac{r^2}{2q}, \qquad q>0,\ -1<\eta<1. \tag{3.2}$$

For t<1, q is uniquely determined by (z,t) in the stated region. The paper gives q comparable to tau+|z|^(1/D). On the central similarity region q is comparable to tau. The leading profiles E,U,V0,Pi determine

$$u_\theta^{(0)}=q^{-A}E(X,\eta), \qquad u_z^{(0)}=q^{-A}U(X,\eta), \qquad r u_r^{(0)}=V_0(X,\eta), \qquad p^{(0)}=q^{-2A}\Pi(X,\eta). \tag{4.3, 4.5}$$

The profiles are selected to make the field Cartesian-smooth at the axis for every t<1. In particular, E=sqrt(2X)F and V0=Xv0 with smooth profile functions. Axisymmetric incompressibility determines V0; the leading radial pressure balance is

$$\frac{1}{r}\partial_r(ru_r^{(0)})+\partial_z u_z^{(0)}=0, \qquad \partial_r p^{(0)}=\frac{(u_\theta^{(0)})^2}{r}. $$

The core scales and characteristic velocities are

$$\ell_r\asymp\tau^{1/2}, \qquad \ell_z\asymp\tau^{1/2-h}, \qquad \frac{\ell_r}{\ell_z}\asymp\tau^h\to0;$$
$$|u_\theta|,|u_z|\asymp\tau^{-1/2-h}, \qquad |u_r|=O(\tau^{-1/2}).$$

The core kinetic energy is of order tau^(1/2-3h), which tends to zero for the paper's h range even though peak velocity diverges. For each fixed nu>0, the paper estimates angular Reynolds number Re_theta asymptotic to tau^(-h), while the radial Reynolds number remains O(1): viscosity remains relevant to radial transport in the collapsing core.

## 4. Stress Cancellation and Oscillatory Pulses

The leading profile is designed so that its tangential momentum residual in the transition annulus can be represented as the cylindrical divergence of a prescribed two-component stress. Proposition 4.2 expresses the leading residual components as negative radial divergences, with the cylindrical factors appropriate to r-theta and r-z stress components. The physical stress has scale q^(-A-1/2) times a profile T0.

Oscillatory velocity pulses are placed in radial/axial patches and extend around complete rings. Their angular means can vanish while their averaged quadratic products do not. In particular, covariance terms such as <w_r w_theta> and <w_r w_z> transport angular and axial momentum. These mean momentum fluxes are arranged to supply the stress missing from the background residual.

The pulses are amplified by background shear. Their nonlinear momentum flux, not a nonzero mean pulse velocity, is the mechanism that acts like an internal stress on the background. Viscosity is included in the construction. The paper describes exponentially small external forcing as seeding pulses; subsequent pulse growth extracts energy from the background shear. Carefully state both roles: the pulses and their correction scheme create the desired residual cancellation, while the final external force is the smooth residual required by the theorem.

### Mechanism chain
1. The leading collapsing vortex satisfies the core balance but leaves a singular residual in the core-to-exterior annulus.
2. Choose the residual stress direction so it lies in a cone realizable by positive wave covariance.
3. Construct high-frequency oscillatory fields whose averaged quadratic velocity products realize that covariance.
4. Correct pressure, nonzero angular harmonics, averaged stresses, and finite moment constraints while preserving incompressibility and support.
5. Iterate the correction scheme so the remaining residual gains decay at each stage and becomes flat at the singular point.

Do not describe the pulses as a generic numerical turbulence model. They are analytically designed waves with prescribed supports, scales, amplitudes, phases/covariances, and correction equations.

## 6. Residual Flatness, Localization, and Viscosity Scaling

The local construction in Theorem 3.1 ensures that for every fixed spatial multi-index alpha, time derivative order b, residual order N, and finite profile radius X1,

$$|\partial_x^\alpha\partial_t^b R(u,p)|\le C_{\alpha,b,N,X_1}q^N \quad (0\le X\le X_1,\ q\downarrow0). \tag{3.4}$$

This means the residual and every fixed Cartesian space-time derivative vanish faster than any prescribed power of q in the central region. The local field is then cut off in space using potentials/coefficients chosen so that the added residual remains smooth. The force is extended across t=1 using compatible one-sided derivative limits, yielding the smooth compactly supported f required by Theorem 1.1. Do not confuse a flat residual with a zero residual everywhere: outside/transition regions and the final force construction matter.

For a viscosity-one construction (u,p,f), the paper rescales to any nu>0 by

$$u_\nu(x,t)=\sqrt{\nu}\,u(x/\sqrt{\nu},t), \qquad p_\nu(x,t)=\nu p(x/\sqrt{\nu},t), \qquad f_\nu(x,t)=\sqrt{\nu}f(x/\sqrt{\nu},t).$$

The singular time is unchanged; the paper verifies the equation and energy scaling. The exterior profile also solves a radial swirl heat equation (equation (3.5)); transcribe it only when discussing that exact exterior construction, and distinguish it from the full nonlinear core equations.
