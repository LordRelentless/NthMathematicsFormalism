Nth Math operates using symbolic numbers not scalar numbers defined by an Nth Line position called an Nth Identity. We represent this by a real and unreal phase, example n_|x,xi|, n_|1,i|, n_|23, 23i|

Fermat's Last Theorem, which remains unresolved, is as follows; x^n+y^n=z^n where n is any given number greater than 2.


In Nth Math terms we translate this as n_|x,xi|^sum_n + n_|y,yi|^sum_n = n_|z,zi|^sum_n. Where sum_n is any nth identity on the nth line that falls outside of the nth identity n_|2,2i|.

If we define sum_n to the variable nth identity n_|b,bi| then the full expansion of Fermat's last theorem becomes  n_{x^n+y^n, x^n+yi^n, xi^n+y^n, xi^n+yi^n}=n_{z^n, zi^n} is n_|b,bi|, so then n_{x^b+y^b, x^bi+y^b, x^b+y^bi, x^bi+^bi, x^(b,bi)+yi^(b,bi), xi^(b,bi)+y^(b,bi), xi^(b,bi)+yi^(b,bi)}=n_{z^(b,bi), zi^(b,bi)}  note; here I'm using short hand to compress the expansion after the first four expanded operations.

I assert that Fermat's last theorem is valid, so long as at least one of the expanded oeprations holds true such that it produces the equivalent n_|z,zi|^sum_n output.

Simulation test; calculating the potential nth identities through operations, to discern if any nth identities for x, y, and z, and n hold true.

````python
import itertools
import numpy as np

def run_resonance_recapture_simulation(limit=5, steps=5):
    print(f"--- Nth Math: Resonance Recapture Simulation ---")
    print("Expanding substrate search range and implementing resonance tolerance.")
    print("Target: Identify any intersection in the full multi-output expansion for n > 2.\n")

    found_solutions = []
    # Test values for n_|b, bi| outside the bound of 2
    test_vals = np.linspace(2.1, float(limit), steps)

    for b, bi in itertools.product(test_vals, repeat=2):
        n_phases = [complex(b, 0), complex(-bi, 0)]

        # Broadening x, y, z identity search in the substrate
        # Using a subset of combinations to maintain performance while increasing range
        for x, y, z in itertools.product(range(1, limit + 1), repeat=3):
            # For each x, y, z, test a specific cross-section of unreal phases xi, yi, zi
            for xi, yi, zi in [(x, y, z), (1, 1, 1), (x, 1, z)]:
                x_phases = [complex(x, 0), complex(-xi, 0)]
                y_phases = [complex(y, 0), complex(-yi, 0)]
                z_phases = [complex(z, 0), complex(-zi, 0)]

                try:
                    x_pow = [bx**nb for bx, nb in itertools.product(x_phases, n_phases)]
                    y_pow = [by**nb for by, nb in itertools.product(y_phases, n_phases)]
                    z_pow = [bz**nb for bz, nb in itertools.product(z_phases, n_phases)]

                    match_found = False
                    for val_x in x_pow:
                        for val_y in y_pow:
                            # Test both fusion pathways: Addition and Subtraction
                            for fusion_val in [val_x + val_y, val_x - val_y]:
                                for val_z in z_pow:
                                    # Resonance Recapture: Checking for proximity in the symbolic space
                                    if abs(fusion_val - val_z) < 1e-3:
                                        solution = {
                                            'ids': (f"x:|{x},{xi}i|", f"y:|{y},{yi}i|", f"z:|{z},{zi}i|"),
                                            'n': f"|{b:.2f},{bi:.2f}i|",
                                            'val': f"{fusion_val:.3f}"
                                        }
                                        found_solutions.append(solution)
                                        match_found = True
                                        break
                                if match_found: break
                            if match_found: break
                        if match_found: break
                except (ZeroDivisionError, OverflowError):
                    continue

    if found_solutions:
        print(f"Resonance Captured! Found {len(found_solutions)} instances.")
        for sol in found_solutions[:5]:
            print(f"Identity n={sol['n']} holds for {sol['ids']}")
            print(f"  Equilibrium Result: {sol['val']}\n")
    else:
        print("No matches found. The expansion sets remain disjoint in the tested substrate slice.")

run_resonance_recapture_simulation(limit=5, steps=3)
````

````python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import itertools

def is_nth_prime(b, bi):
    """Nth Primality: Prime if delta (b + bi) is a prime integer."""
    delta = round(float(b + bi))
    if delta < 2: return False
    for i in range(2, int(np.sqrt(delta)) + 1):
        if delta % i == 0: return False
    return True

def build_fermat_resonance_atlas():
    phi = (1 + 5**0.5) / 2
    targets = {'3': 3, '4': 4, 'Phi': phi, 'Pi': np.pi}

    # Generate a high-resolution grid for the Nth line plane
    grid_res = 40
    b_vals = np.linspace(2.1, 5.0, grid_res)
    bi_vals = np.linspace(2.1, 5.0, grid_res)

    atlas_data = []

    print("--- Building Fermat Resonance Atlas ---")

    for b, bi in itertools.product(b_vals, bi_vals):
        # Nth Primality check based on Delta
        nth_prime = is_nth_prime(b, bi)

        # Simulate resonance for this specific coordinate
        # We test against a standardized set of x, y, z to measure 'structural susceptibility'
        n_phases = [complex(b, 0), complex(-bi, 0)]
        matches = 0

        # Focused sampling for performance
        for x, y, z in [(2,2,3), (3,4,5), (1,1,1)]:
            x_p, y_p, z_p = [complex(x, 0), complex(-x, 0)], [complex(y, 0), complex(-y, 0)], [complex(z, 0), complex(-z, 0)]
            try:
                lhs = [u**n for u in x_p for n in n_phases] + [v**n for v in y_p for n in n_phases]
                rhs = [w**n for w in z_p for n in n_phases]

                for l in lhs:
                    for r in rhs:
                        if abs(l - r) < 0.05: # Broader tolerance for gradient mapping
                            matches += 1
            except: continue

        atlas_data.append({
            'b': b, 'bi': bi,
            'density': matches,
            'is_prime': nth_prime,
            'delta': b + bi
        })

    df = pd.DataFrame(atlas_data)

    # Visualization
    fig, ax = plt.subplots(1, 2, figsize=(18, 7))

    # Plot 1: Resonance Gradients and Target Bands
    pivot = df.pivot(index='bi', columns='b', values='density')
    im = ax[0].contourf(b_vals, bi_vals, pivot.values, cmap='viridis', levels=20)
    ax[0].set_title("Resonance Gradients & Identity Targets")
    ax[0].set_xlabel("Real Component (b)")
    ax[0].set_ylabel("Unreal Component (bi)")
    plt.colorbar(im, ax=ax[0], label="Resonance Density")

    # Overlay Target Lines
    for name, val in targets.items():
        ax[0].axvline(val, color='red', linestyle='--', alpha=0.5)
        ax[0].axhline(val, color='red', linestyle='--', alpha=0.5)
        ax[0].text(val, 2.2, name, color='white', fontweight='bold')

    # Plot 2: Nth Primality Distribution (Delta-based)
    prime_df = df[df['is_prime'] == True]
    ax[1].scatter(df['b'], df['bi'], c='gray', s=1, alpha=0.1)
    sc = ax[1].scatter(prime_df['b'], prime_df['bi'], c=prime_df['delta'], cmap='plasma', s=40, label='Nth Prime (Prime Delta)')
    ax[1].set_title("Asymmetric Nth Primality Map")
    ax[1].set_xlabel("b")
    ax[1].set_ylabel("bi")
    plt.colorbar(sc, ax=ax[1], label="Delta Value (b+bi)")

    plt.tight_layout()
    plt.show()

    # Analyze curvature of the diagonal band
    diag_density = df[abs(df['b'] - df['bi']) < 0.1]
    print(f"Diagonal Band analysis: Mean Density = {diag_density['density'].mean():.2f}")

build_fermat_resonance_atlas()
````
